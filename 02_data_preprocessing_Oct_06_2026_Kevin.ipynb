```python
import pandas as pd

df = pd.read_csv("/Users/kevinvillon/Desktop/NOVA_IT691/data/cicids2017_cleaned_WHOLE_SET.csv")
#please remember to change the path within the quotes corresponding to where the file is in your device

print(df.shape)
```

```python
print(df.columns.tolist())

target_column = "Attack Type"

X = df.drop(columns=[target_column])
y = df[target_column]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.15,
    random_state=42,
    stratify=y
)
```

```python
validation_size = 0.15 / 0.85

X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=validation_size,
    random_state=42,
    stratify=y_train_val
)
```

```python
#Verifying if dataset split into three sets by checking sample sizes

print("Full dataset:", len(df))
print("Training set:", len(X_train))
print("Validation set:", len(X_val))
print("Test set:", len(X_test))
```

```python
print("\nTraining percentage:")
print(len(X_train) / len(df) * 100)

print("\nValidation percentage:")
print(len(X_val) / len(df) * 100)

print("\nTest percentage:")
print(len(X_test) / len(df) * 100)
```

```python
print("TRAINING")
print(y_train.value_counts(normalize=True) * 100)

print("\nVALIDATION")
print(y_val.value_counts(normalize=True) * 100)

print("\nTEST")
print(y_test.value_counts(normalize=True) * 100)
```

```python
train_df = X_train.copy()
train_df["Attack Type"] = y_train

val_df = X_val.copy()
val_df["Attack Type"] = y_val

test_df = X_test.copy()
test_df["Attack Type"] = y_test
```

```python
train_df.to_csv("training_dataset_CIC_70.csv", index=False)
val_df.to_csv("validation_dataset_CIC_15.csv", index=False)
test_df.to_csv("initial_test_dataset_CIC_15.csv", index=False)
```
